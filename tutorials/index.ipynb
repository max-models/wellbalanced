# Tutorials

Jupyter notebooks in this section are executed when the documentation is built. Download a notebook from its page to run it yourself.

- [A global model in five steps](/carafe/tutorials/global-model/): an argon discharge, its balances and the closed form of the textbook model.
- [Sources and gas heating](/carafe/tutorials/sources-and-gas/): external volume sources such as a beam's, and the gas temperature as a balance.
- [1D boundaries](/carafe/tutorials/fluid-1d/): `Fluid1D` with ions with momentum, the kinetic wall and an outflow.
- [OpenCR reaction sets](/carafe/tutorials/reaction-sets/): hydrogen's effective rates from collisional-radiative models (`carafe.cr`).
- [Beam neutralizer](/carafe/tutorials/neutralizer/): a beam's charge states and the plasma it makes.
- [RF driver](/carafe/tutorials/rf-driver/): from the 0D driver to the inductive field in r–z and a 2D fluid.
- [Ionization region model](/carafe/tutorials/ionization-region/): the IRM of a HiPIMS pulse, its outputs and the fit of $\beta_t$ and $f$.
- [3D inductive field](/carafe/tutorials/field-3d/): the field on the edges of a 3D grid, checked on a solenoid.
- [Spoke model](/carafe/tutorials/spoke-model/): the 2D drift fluid of a magnetron and `spoke_report`.
- [Metal and self-sputtering](/carafe/tutorials/metal-species/): a sputtered metal target in the drift fluid: energy-dependent yields, singly and doubly charged metal ions and the cathode current by species.
- [Hall thruster 1D](/carafe/tutorials/hall-thruster/): the axial model of an SPT-100-like thruster, its profiles, its discharge current and a closure of the anomalous transport.
- [MHD on patches](/carafe/tutorials/mhd/): ideal MHD on a patchwork hierarchy: the Orszag–Tang vortex, the order on an Alfvén wave and a loop across a refined patch.
- [MHD test problems](/carafe/tutorials/mhd-test-problems/): the rotor, the magnetized blast wave and the energy budget of the Orszag–Tang vortex.
- [MHD in 3D](/carafe/tutorials/mhd-3d/): `MHD3D`, the vector potential on the edges: the order on an oblique Alfvén wave, the 2D limit and a 3D blast wave.
- [MHD performance](/carafe/tutorials/mhd-performance/): where the time of an `MHD2D` step goes, with scope-profiler's `carafe` tag.
