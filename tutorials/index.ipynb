# Tutorials

Jupyter notebooks in this section are executed when the documentation is built. Download a notebook from its page to run it yourself.

- [A global model in five steps](/wellbalanced/tutorials/global-model/): an argon discharge, its balances and the closed form of the textbook model.
- [Sources and gas heating](/wellbalanced/tutorials/sources-and-gas/): external volume sources such as a beam's, and the gas temperature as a balance.
- [1D boundaries](/wellbalanced/tutorials/fluid-1d/): `Fluid1D` with ions with momentum, the kinetic wall and an outflow.
- [OpenCR reaction sets](/wellbalanced/tutorials/reaction-sets/): hydrogen's effective rates from collisional-radiative models (`wellbalanced.cr`).
- [Beam neutralizer](/wellbalanced/tutorials/neutralizer/): a beam's charge states and the plasma it makes.
- [RF driver](/wellbalanced/tutorials/rf-driver/): from the 0D driver to the inductive field in r–z and a 2D fluid.
- [3D inductive field](/wellbalanced/tutorials/field-3d/): the field on the edges of a 3D grid, checked on a solenoid.
- [Spoke model](/wellbalanced/tutorials/spoke-model/): the 2D drift fluid of a magnetron and `spoke_report`.
